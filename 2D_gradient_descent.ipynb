```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x, y):
    return x**2 + y**2

# Gradient
def grad(x, y):
    return np.array([2*x, 2*y])
```

```python
# Initialize
point = np.array([5.0, 5.0])
lr = 0.1
iterations = 30

points = [point.copy()]

# Gradient descent loop
for _ in range(iterations):
    point = point - lr * grad(point[0], point[1])
    points.append(point.copy())

points = np.array(points)
```

```python
# Create grid
x = np.linspace(-6, 6, 100)
y = np.linspace(-6, 6, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Plot contour
plt.figure()
plt.contour(X, Y, Z, levels=20)

# Plot path
plt.plot(points[:,0], points[:,1], marker='o')

plt.title("2D Gradient Descent Path")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.show()
```
