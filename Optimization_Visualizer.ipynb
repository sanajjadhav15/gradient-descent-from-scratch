```python
import numpy as np

def f(x,y):
    return x**2+2*y**2

def grad(x,y):
    return np.array([2*x,4*y])
```

```python
def gradient_descent_2d(start,lr,steps):
    point=np.array(start)
    path=[point.copy()]

    for _ in range(steps):
        g=grad(point[0],point[1])
        point=point-lr*g
        path.append(point.copy())

    return np.array(path)
```

```python
def momentum_descent_2d(start,lr,beta,steps):
    point=np.array(start)
    v=np.zeros_like(point)

    path=[point.copy()]

    for _ in range(steps):
        g=grad(point[0],point[1])
        v=v*beta+g
        point=point-lr*v
        path.append(point.copy())

    return np.array(path)
```

```python
import matplotlib.pyplot as plt

x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)

X,Y=np.meshgrid(x,y)
Z=f(X,Y)

plt.figure()
plt.contour(X,Y,Z,levels=20)
```

```python
path1 = gradient_descent_2d([4, 4], 0.1, 20)
path2 = momentum_descent_2d([4, 4], 0.1, 0.9, 20)

plt.plot(path1[:,0], path1[:,1], marker='o', label="Gradient Descent")
plt.plot(path2[:,0], path2[:,1], marker='x', label="Momentum")

plt.legend()
plt.title("Optimization Paths")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```
