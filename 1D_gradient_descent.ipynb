```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define function and its derivative
def f(x):
    return x**2

def grad(x):
    return 2*x
```

```python
# 2. Gradient Descent
x = 5           # starting point
lr = 0.1             # learning rate
iterations = 20

x_values = [x]
y_values = [f(x)]

for i in range(iterations):
    x = x - lr * grad(x)
    x_values.append(x)
    y_values.append(f(x))
```

```python
# 3. Print progression
print("Iteration-wise x values:")
for i, val in enumerate(x_values):
    print(f"Step {i}: x = {val:.4f}, f(x) = {f(val):.4f}")


# 4. Plot function curve
x_range = np.linspace(-6, 6, 100)
y_range = f(x_range)

plt.figure()
plt.plot(x_range, y_range)  # function curve
plt.scatter(x_values, y_values)  # descent points

# Connect the points
plt.plot(x_values, y_values)

plt.title("Gradient Descent on f(x) = x^2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```
